```python
import requests

def get_live_weather(city_name):
    print(f"Fetching live data for {city_name}...")
    
    # Step 1: City ke naam se uske Latitude aur Longitude nikalna (Geocoding)
    geo_url = f"https://geocoding-api.open-meteo.com/v1/search?name={city_name}&count=1"
    geo_response = requests.get(geo_url).json()
    
    if not geo_response.get("results"):
        print("Error: City not found. Please check the spelling.")
        return None, None
        
    lat = geo_response["results"][0]["latitude"]
    lon = geo_response["results"][0]["longitude"]
    
    # Step 2: Un coordinates ka use karke live Temperature aur Humidity lana
    weather_url = f"https://api.open-meteo.com/v1/forecast?latitude={lat}&longitude={lon}&current=temperature_2m,relative_humidity_2m"
    weather_response = requests.get(weather_url).json()
    
    current_weather = weather_response["current"]
    temp = current_weather["temperature_2m"]
    humidity = current_weather["relative_humidity_2m"]
    
    print("\n--- Live Weather API Result ---")
    print(f"City: {city_name.capitalize()}")
    print(f"Temperature: {temp} °C")
    print(f"Humidity: {humidity} %")
    print("-------------------------------\n")
    
    return temp, humidity

# Test karne ke liye input lein
city = input("Apni city ka naam enter karein (e.g., Chandigarh, Patiala): ")
get_live_weather(city)
```
